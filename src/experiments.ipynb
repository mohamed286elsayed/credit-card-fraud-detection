```python
import pandas as pd
from  sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
import numpy as np
```

```python
train_df = pd.read_csv('C:/Users/USER/Desktop/credit_card_v2/data/train.csv')
val_df = pd.read_csv('C:/Users/USER/Desktop/credit_card_v2/data/val.csv')
```

```python
# try data as it is  --> 0.77
# extract hour --> 0.76
# remove time --> 0.75

X_train = train_df.drop(columns=['Class'])
y_train = train_df['Class']


X_val = val_df.drop(columns=['Class'])
y_val = val_df['Class']


```

```python
lr = LogisticRegression(max_iter= 2000, random_state= 42)
lr.fit(X_train, y_train)

y_pred = lr.predict(X_val)
print(classification_report(y_val, y_pred))
```

```python
X_train['Hour'] = (X_train['Time'] // 3600) % 24
X_val['Hour'] = (X_val['Time'] // 3600) % 24

lr = LogisticRegression(max_iter= 2000, random_state= 42)
lr.fit(X_train, y_train)

y_pred = lr.predict(X_val)
print(classification_report(y_val, y_pred))
```

```python
# remove time
X_train = X_train.drop(columns=['Time'])
X_val = X_val.drop(columns=['Time'])

lr = LogisticRegression(max_iter= 2000, random_state= 42)
lr.fit(X_train, y_train)

y_pred = lr.predict(X_val)
print(classification_report(y_val, y_pred))
```

```python
X_train = X_train.drop(columns=['Hour'])
X_val = X_val.drop(columns=['Hour'])

X_train['Amount'] = np.log1p(X_train['Amount'])
X_val['Amount'] = np.log1p(X_val['Amount'])



lr = LogisticRegression(max_iter= 2000, random_state= 42)
lr.fit(X_train, y_train)

y_pred = lr.predict(X_val)
print(classification_report(y_val, y_pred))
```

```python
from sklearn.preprocessing import RobustScaler
scaler = RobustScaler()

X_train['Amount'] = np.log1p(X_train['Amount'])
# restore Time if it was removed earlier
if 'Time' not in X_train.columns:
    X_train['Time'] = train_df['Time']
if 'Time' not in X_val.columns:
    X_val['Time'] = val_df['Time']

# keep Amount transformed consistently in both splits
X_train['Amount'] = np.log1p(X_train['Amount'])
X_val['Amount'] = np.log1p(X_val['Amount'])

# fit the scaler on train and transform val with the same scaler
X_train['Time'] = scaler.fit_transform(X_train[['Time']]).ravel()
X_val['Time'] = scaler.transform(X_val[['Time']]).ravel()
X_val['Time'] = scaler.transform(X_val[['Time']]).ravel()


```
